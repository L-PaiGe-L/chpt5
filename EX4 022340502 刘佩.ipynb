# 子图布局综合可视化分析

基于五个MATLAB文件的子图布局可视化，涵盖多种子图布局技术

## 第1题：基于`plt.subplot()`的多子图绘制

通过子图布局展示3类数据：智能手表季度销量、蓝牙耳机季度销量、各平台男女用户数分布

```python
import matplotlib.pyplot as plt
import numpy as np

# 数据准备
quarters = ['第一季度', '第二季度', '第三季度', '第四季度']
watch_sales = [50, 75, 125, 100]  # 智能手表销量
headphone_sales = [190, 210, 240, 220]  # 蓝牙耳机销量

platforms = ['知乎', '小红书', '微博', '抖音']
male_users = [20, 10, 50, 70]  # 男性用户数（百万）
female_users = [15, 50, 60, 80]  # 女性用户数（百万）

# 创建2行2列子图布局
plt.figure(figsize=(12, 8))

# 子图1：智能手表季度销量（左上）
plt.subplot(2, 2, 1)
plt.plot(range(1, 5), watch_sales, color='#1f77b4', linewidth=2, marker='o', markersize=6)
plt.xlabel('季度')
plt.ylabel('销量')
plt.title('智能手表季度销量')
plt.xticks(range(1, 5), quarters)
plt.ylim(0, 175)
plt.grid(True)

# 子图2：蓝牙耳机季度销量（右上）
plt.subplot(2, 2, 2)
plt.plot(range(1, 5), headphone_sales, color='#ff7f0e', linewidth=2, marker='s', markersize=6)
plt.xlabel('季度')
plt.ylabel('销量')
plt.title('蓝牙耳机季度销量')
plt.xticks(range(1, 5), quarters)
plt.ylim(180, 250)
plt.grid(True)

# 子图3：各平台男女用户数分布（下方合并2列）
plt.subplot(2, 2, (3, 4))
x = np.arange(len(platforms))
width = 0.35

plt.bar(x - width/2, male_users, width, label='男性用户(百万)', color='#2ca02c')
plt.bar(x + width/2, female_users, width, label='女性用户(百万)', color='#d62728')

plt.xlabel('平台')
plt.ylabel('用户数（百万）')
plt.title('各平台男女用户分布')
plt.xticks(x, platforms)
plt.ylim(0, 80)
plt.legend(loc='upper left')
plt.grid(True)

plt.suptitle('设备季度销量与平台用户性别分布')
plt.tight_layout()
plt.show()
```

## 第2题：基于`plt.subplots()`的并列子图绘制

通过1行2列的并列子图展示两组月度总访问量数据

```python
# 数据准备
months = ['一月', '二月', '三月', '四月']
visits1 = [0.5, 1.2, 2.0, 1.8]  # 第一组月度访问量
visits2 = [0.8, 1.5, 1.9, 2.2]  # 第二组月度访问量

# 创建1行2列并列子图
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# 子图1：第一组月度总访问量
ax1.plot(range(1, 5), visits1, color='#9467bd', linewidth=2, marker='^', markersize=6)
ax1.set_xlabel('月份')
ax1.set_ylabel('访问量')
ax1.set_title('月度总访问量（组1）')
ax1.set_xticks(range(1, 5))
ax1.set_xticklabels(months)
ax1.set_ylim(0, 2.5)
ax1.grid(True)

# 子图2：第二组月度总访问量
ax2.plot(range(1, 5), visits2, color='#8c564b', linewidth=2, marker='v', markersize=6)
ax2.set_xlabel('月份')
ax2.set_ylabel('访问量')
ax2.set_title('月度总访问量（组2）')
ax2.set_xticks(range(1, 5))
ax2.set_xticklabels(months)
ax2.set_ylim(0, 2.5)
ax2.grid(True)

plt.suptitle('月度总访问量对比（五月、六月数据待补充）')
plt.tight_layout()
plt.show()
```

## 第3题：基于`plt.subplot2grid()`的网格布局子图绘制

通过2行1列的网格布局展示两组月度总访问量数据

```python
# 数据准备（新趋势数据，避免与第2题重复）
months = ['一月', '二月', '三月', '四月']
visits3 = [1.0, 1.8, 2.2, 1.5]  # 第三组月度访问量
visits4 = [0.6, 1.3, 2.1, 1.9]  # 第四组月度访问量

# 创建2行1列网格布局
plt.figure(figsize=(8, 10))

# 子图1：第一行网格（上方）
plt.subplot(2, 1, 1)
plt.plot(range(1, 5), visits3, color='#e377c2', linewidth=2, marker='d', markersize=6)
plt.xlabel('月份')
plt.ylabel('访问量')
plt.title('月度总访问量（上）')
plt.xticks(range(1, 5), months)
plt.ylim(0, 2.5)
plt.grid(True)

# 子图2：第二行网格（下方）
plt.subplot(2, 1, 2)
plt.plot(range(1, 5), visits4, color='#7f7f7f', linewidth=2, marker='p', markersize=6)
plt.xlabel('月份')
plt.ylabel('访问量')
plt.title('月度总访问量（下）')
plt.xticks(range(1, 5), months)
plt.ylim(0, 2.5)
plt.grid(True)

plt.suptitle('月度总访问量网格布局（五月、六月数据待补充）')
plt.tight_layout()
plt.show()
```

## 第4题：基于`ax.twinx()`的双轴图绘制

通过双Y轴图表同时展示股价与成交量的月度变化

```python
# 数据准备
months = ['一月', '二月', '三月', '四月', '五月', '六月']
stock_price = [105, 110, 120, 115, 130, 125]  # 股价数据
volume = [0.8, 1.2, 2.5, 1.8, 2.8, 2.0]       # 成交量数据

# 创建双轴图
fig, ax1 = plt.subplots(figsize=(10, 6))

# 左侧Y轴：股价
ax1.plot(range(1, 7), stock_price, color='blue', linewidth=2, marker='o', markersize=6, label='股价')
ax1.set_xlabel('月份')
ax1.set_ylabel('股价', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_ylim(100, 135)
ax1.set_yticks(range(100, 136, 5))

# 右侧Y轴：成交量
ax2 = ax1.twinx()
ax2.plot(range(1, 7), volume, color='red', linewidth=2, marker='s', markersize=6, label='成交量')
ax2.set_ylabel('成交量', color='red')
ax2.tick_params(axis='y', labelcolor='red')
ax2.set_ylim(0, 3.0)
ax2.set_yticks([i * 0.5 for i in range(7)])

# 通用设置
plt.xticks(range(1, 7), months)
plt.title('股价与成交量趋势')

# 合并图例
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.grid(True)
plt.tight_layout()
plt.show()
```

## 第5题：基于`gridspec()`的综合子图布局绘制

通过多子图组合展示年度数据趋势与各学院招生统计

```python
import matplotlib.gridspec as gridspec

# 数据准备
years = [2021, 2022, 2023, 2024]
year_data = [20000, 21500, 22800, 23500]  # 2021-2024趋势数据

colleges = ['工学院', '文学院', '商学院', '理学院']
enrollment = [7000, 3000, 6000, 4000]     # 招生人数
proportion = [34.8, 21.7, 26.1, 17.4]     # 招生占比（%）

# 创建网格布局
fig = plt.figure(figsize=(12, 8))
gs = gridspec.GridSpec(2, 2, figure=fig)

# 子图1：2021-2024年度趋势（上半部分合并）
ax1 = fig.add_subplot(gs[0, :])
ax1.plot(years, year_data, color='darkgreen', linewidth=3, marker='o', markersize=8)
ax1.set_xlabel('年份')
ax1.set_ylabel('数值')
ax1.set_title('2021-2024年度数据趋势')
ax1.set_xticks(years)
ax1.set_ylim(20000, 24000)
ax1.set_yticks(range(20000, 24001, 500))
ax1.grid(True)

# 子图2：各学院招生人数（左下）
ax2 = fig.add_subplot(gs[1, 0])
ax2.bar(colleges, enrollment, color='skyblue', edgecolor='black')
ax2.set_xlabel('学院')
ax2.set_ylabel('招生人数')
ax2.set_title('各学院招生人数')
ax2.set_ylim(0, 8000)
ax2.set_yticks(range(0, 8001, 1000))
ax2.tick_params(axis='x', rotation=45)

# 子图3：各学院招生占比（右下）
ax3 = fig.add_subplot(gs[1, 1])
ax3.pie(proportion, labels=colleges, autopct='%1.1f%%', colors=plt.cm.Set3.colors)
ax3.set_title('各学院招生占比')

plt.suptitle('年度数据与各学院招生统计')
plt.tight_layout()
plt.show()
```
